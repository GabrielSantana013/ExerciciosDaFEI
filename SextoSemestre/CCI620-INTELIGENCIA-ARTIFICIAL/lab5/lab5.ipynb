```python
import pandas as pd
df = pd.read_csv("/content/notas.csv")
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np
```

```python

x = []
for i in df.values:
  x.append([i[0] ,i[1]])
print(x)
x1 = [x[0] for x in df.values]
x2 = [x[1] for x in df.values]
y = [x[2] for x in df.values]

fig = plt.figure()
ax = fig.add_subplot(projection='3d')

ax.scatter(x1, x2, y, cmap='viridis', s=100, edgecolors='k')

ax.set_xlabel('Livro')
ax.set_ylabel('Frequencia')
ax.set_zlabel('Nota')
```

```python
#Divide o dataset em treino e teste
X_train, X_test, Y_train, Y_test = train_test_split(x,y, test_size=0.2, random_state=42)

#Faz a regressao
modelo = LinearRegression()
modelo.fit(X_train,Y_train)

#2 livros e 5 aulas
print(modelo.predict([[2,11]]))

print(modelo.predict([[0,5]]))

print(modelo.predict([[4,20]]))

print(modelo.predict([[2,10]]))

print(modelo.predict([[4,15]]))


```

```python
df = pd.read_csv("/content/agua_alpes.csv")

X = df[["BPt"]]       # 2D
y = df["Pressure"]     # 1D

plt.scatter(x, y)
plt.xlabel("bpt")
plt.ylabel("pressao")

#Faz a regressao
modelo = LinearRegression()
modelo.fit(X, y)

#pressao para 200°
print(modelo.predict([[200]]))

#pressao para 215°
print(modelo.predict([[215]]))
```

```python
a = modelo.coef_[0]
b = modelo.intercept_

print("Coef Angular:", a)
print("Coef Linear:", b)

# Previsões para todos os X
previsao = modelo.predict(X)

# Gráfico
plt.scatter(X, y, color="red", label="Dados")

# Ordenar X para a reta não ficar "ziguezagueando"
plt.plot(X, previsao, color="blue", label="Regressão Linear")

plt.xlabel("BPt")
plt.ylabel("Pressure")
plt.legend()
plt.show()
```
