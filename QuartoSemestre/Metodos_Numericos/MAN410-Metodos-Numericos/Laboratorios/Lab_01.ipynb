# Questão 1


```python
import numpy as np
import sympy as sp
```

```python
x = sp.symbols('x')

expr = (x**2-2*x)/(sp.cos(x))
```

```python
def f(x):
    return (x**2-2*x)/np.cos(x)
```

```python
# 1a
print("usando numpy: ", f(1))
print("usando sympy: ", expr.subs(x,1).evalf())
```

```python
# 1b
print(sp.limit(expr,x,9).evalf())
```

```python
# 1c
print(sp.diff(expr,x))
```

## Questão 2

```python
import numpy as np

A = np.array([[1,0,1],[1,2,3],[1,2,4]])
B = np.array([[3,1,4], [1,2,1], [3,1,1]])
```

```python
#2a
det_A = np.linalg.det(A)
print(det_A)
```

```python
#2b
inv_B = np.linalg.inv(B)
print(inv_B)
```

```python
#2c
trans_A = A.T
print(trans_A)
```

```python
#2d
mult_AB = np.dot(A,B)
print(mult_AB)
```

```python
#2e
new_B = -4*B
resp = A+new_B
print(resp)
```
