## Questão 1 (segunda atividade)



```python
from sympy import *
x = symbols('x')
eq=sqrt(x**2+1)
#calculo da deriva terceira
r=diff(eq,x,3)
print(r)
#calcular o valor da derivada para x=3
z=r.subs(x,3)
print('='*20)
print(z.evalf(2))
```

## Questão 2 

```python
import numpy as np

eq = np.matrix([[-1,1,-1],[1,2,4],[3,1,-2]])
resp = np.matrix([[5],[4],[-3]])

r = np.linalg.solve(eq,resp)
print("Resposta: ", r)

print('='*20)

#cria a matriz incompleta
matriz_D=np.matrix([[-1,1,-1],[1,2,4],[3,1,-2]])
#calcula o determinante de D
D=np.linalg.det(matriz_D)
print(round(D,2))
#cria a matriz Dx
matriz_Dx=np.matrix([[5,1,-1],[4,2,4],[-3,1,-2]])
Dx=np.linalg.det(matriz_Dx)
#cria a matriz Dy
matriz_Dy=np.matrix([[-1,5,-1],[1,4,4],[3,-3,-2]])
Dy=np.linalg.det(matriz_Dy)
#cria a matriz Dz
matriz_Dz=np.matrix([[-1,1,5],[1,2,4],[3,1,-3]])
Dz=np.linalg.det(matriz_Dz)
x=Dx/D
y=Dy/D
z=Dz/D
print("x=",np.round(x,2))
print("y=",np.round(y,2))
print("z=", np.round(z,2))
```
