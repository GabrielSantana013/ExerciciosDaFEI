Calcular a maior raiz pelo método de Newton Rapshon  ou pela  função fsolve da seguinte equação:

x2+sen(x)-5=0

Coloque a resposta correta no campo indicado.

Usar vírgula na resposta.

Usar duas casas decimais.

```python
import numpy as np
from sympy import *

x=Symbol('x')

eq = (x**2+sin(x)-5)

dfx = diff(eq,x,1)
x_val = 0.5
for i in range (0,6):
    #substitui o valor de x escolhido na eq
    f_val = eq.subs(x,x_val)
    df_val = dfx.subs(x,x_val)
    x_val = x_val-(f_val/df_val)
    print(f"Iteração {i+1}: x = {x_val.evalf()}")
    
```
