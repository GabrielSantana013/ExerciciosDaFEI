a) calcular o coeficiente angular a ou a1 da reta. Usar duas casas decimais.
b) calcular o coeficiente linear b ou a0 da reta. Usar duas casas decimais. 

```python
import numpy as np

x = np.array([1940,1950,1960,1970,1975,1980])
y = np.array([800, 1080, 2000, 2600, 3000, 3600])

n = np.size(x)
Sx= np.sum(x)
Sy=np.sum(y)
Sxy=np.sum(x*y)
Sxx=np.sum(x*x)
a1=(n*Sxy-Sx*Sy)/(n*Sxx-Sx**2)   #Calcula o coeficiente a1 da reta
a0=(Sxx*Sy-Sxy*Sx)/(n*Sxx-Sx**2) #Calcula o coeficiente a0 da reta
print(f'{a1:.2f}')
print(f'{a0:.2f}')
```

c) Calcular o resíduo do ajuste. Usar duas casas decimais.

```python
#funcao que calcula o RS
def Residuo(x,y,b0,b1):
    n = len(y)
    RS = 0
    for i in range(0,n):
         y_pred=b0+b1*x[i]
         RS = RS + (y[i]-y_pred)**2
    return RS
print(f'RS: {Residuo(x,y,a0,a1):.2f}')
```

d) determinar o valor de y para x=1990. Usar duas casas decimais.

```python
# mostra os dados
import matplotlib.pyplot as plt

x = 1990
y_pred = a0 + a1*x

# mostra a reta de regressão
plt.scatter(x, y_pred, color = "r")

plt.xlabel('x', fontsize = 15)
plt.ylabel('y', fontsize = 15)
plt.show(True)

print(f'{y_pred:.2f}')


```

e) determinar o coeficiente de correlação linear. Usar três casas decimais.

```python
correlacao = np.corrcoef(x, y)
print(correlacao)
```

f) calcular o coeficiente de determinação R2. Usar três casas decimais.

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

x = np.array(x)
x = x.reshape(-1, 1)
modelo = LinearRegression().fit(x, y)
R2 = r2_score(y, modelo.predict(x))
print('R2:', R2)
```
